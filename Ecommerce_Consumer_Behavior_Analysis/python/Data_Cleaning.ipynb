```python
# selecting features
```

```python
features= df[['Purchase_Amount','Frequency_of_Purchase']]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
Scaled_features= scaler.fit_transform(features)
```

```python
from sklearn.cluster import KMeans

inertia =[]

for k in range(1,10):
    kmeans = KMeans(n_clusters=k,random_state=42)
    kmeans.fit(Scaled_features)
    inertia.append(kmeans.inertia_)

plt.plot(range(1,10),inertia,marker='o')
plt.title('Elbow Methid')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
plt.show()
```

```python
kmeans = KMeans(n_clusters=3,random_state=42)
df['Cluster']= kmeans.fit_predict(Scaled_features)
```

```python
sns.scatterplot(
    x='Purchase_Amount',
    y='Frequency_of_Purchase',
    hue='Cluster',
    data=df,
    palette='Set1'
)
plt.title('Customer Segmentation')
plt.show
```

```python
df.groupby('Cluster')[['Purchase_Amount', 'Frequency_of_Purchase']].mean()
```

```python
def label_cluster(row):
    if row['Cluster'] == 0:
        return "Low Value"
    elif row['Cluster'] == 1:
        return "High Value"
    else:
        return "Frequent Buyers"

df['Customer_Segment'] = df.apply(label_cluster, axis=1)
```
